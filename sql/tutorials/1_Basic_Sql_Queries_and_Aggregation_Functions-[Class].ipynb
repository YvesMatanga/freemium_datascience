<h4>D. Conditional flow functions </h4>

<h5>IIF(test, true_outcome, false_outcome)</h5>

```python
%%sql

SELECT Total, IIF(Total > 2, 'expensive','cheap') AS Category FROM invoices
```

<h5>CASE (test, option1, option2,...default)</h5>

```python
%%sql
SELECT Name, 
CASE Name 
  WHEN 'MPEG audio file' THEN 'MPEG'
  WHEN 'AAC audio file' THEN 'AAC'
  ELSE 'None'
END AS 'Type'
FROM media_types
```

<h5>IFNULL(is_null_condition, value)</h5>

```python
%%sql
SELECT IFNULL(BillingState,'Empty') FROM invoices
```

<h5>NULLIF()</h5>

```python
%%sql 
SELECT NULLIF(5,5)
```
