<h2>B. Join Operators</h2>

<b style="color:blue"> Join Operators are SQL functionalities to fetch data from multiple tables</b>

Types of Join Operators

<ol>
<li style="color:blue">INNER JOIN</li>
<li  style="color:blue">LEFT JOIN</li>
<li  style="color:blue">RIGHT JOIN</li>
<li  style="color:blue">OUTER JOIN</li>
<li  style="color:blue">CROSS JOIN</li>   
</ol>

<h3>1. Inner Join</h3>

<b style="color:blue">Returns only matching records from both tables.</b>
<img src="../media/inner_join.png" width="500"/>


```python
%%sql
SELECT * FROM artists LIMIT 5
```

```python
%%sql

SELECT * FROM albums LIMIT 5
```

```python
%%sql
SELECT artists.Name AS 'artist', albums.Title AS 'album' FROM artists JOIN albums ON artists.ArtistId = albums.ArtistId LIMIT 10
```

<h3>2. Left Join</h3>
<br/>
<b style="color:blue"> Returns all records from the left table</b> and matching records from the right table.
<img src="../media/left_join.png" width="500"/>


```python
%%sql 

SELECT Artists.Name AS 'artist', albums.Title As 'album' FROM artists LEFT JOIN
albums ON artists.artistid = albums.artistid 
```

<h3>3. RIGHT Join</h3>

<b style="color:blue"> Returns all records from the right table</b> and matching records from the left table.

```python
%%sql


SELECT albums.Title AS 'album', artists.Name As 'artist' FROM albums LEFT JOIN artists ON artists.artistid = albums.artistid
LIMIT 10
    
```

<h3>4. Outer Join</h3>

<b style = "color:blue">Returns all records from both tables.</b>
<b style="color:blue">If there is no match, NULL values appear on the missing side.</b>

<img src="../media/outer_join.png" width="500"/>


```python
%%sql


SELECT artists.Name as 'artist', albums.Title As 'album' FROM artists LEFT JOIN albums ON artists.artistid = albums.artistid

UNION

SELECT artists.Name as 'artist', albums.Title As 'album' FROM albums LEFT JOIN artists ON artists.artistid = albums.artistid

LIMIT 5
```

<h3>5. Cross Join</h3><br/>
 <b style="color:blue">Returns the Cartesian product of both tables (every combination of rows).</b>
 <b>No ON condition is needed.</b>
 
<img src="../media/cross_join.png" width="500"/>


```python
%%sql
SELECT * FROM media_types
```

```python
%%sql

SELECT genres.Name , media_types.Name  AS' media_type'
FROM genres
CROSS JOIN
media_types
LIMIT 10
```

<h3>Set Operator Rules</h3>

<ul>
    <li>The number of columns in both tables needs to be equal.</li>
    <li>The columns from each table that we want to combine must contain compatible datatypes.</li>
    <li>We can only apply the ORDER BY clause to the combined (i.e. UNIONised) table and not to the individual tables</li>
    <li>The GROUP BY clause can only be applied to the individual tables (i.e. before the UNION operation) and not the combined result</li>
</ul>

Types of Set operators:

<ul>
<li style="color:blue">UNION</li>
<li  style="color:blue">INTERSECT</li>
<li  style="color:blue">EXCEPT</li>
</ul>

<h2>1. Union Operator</h2>

<img src="../media/set_union.png" width="500"/>

```python
%%sql
SELECT City, Region, Address FROM Customer LIMIT 5
```

```python
%%sql
SELECT City, Region, Address FROM Supplier LIMIT 10
```

```python
%%sql

SELECT City, Region, Address FROM Customer

UNION 

SELECT City, Region, Address FROM Supplier 

LIMIT 100
```

<h2>2. Intersection Operator</h2>


<img src="../media/intersect.png" width="500"/>

```python
%%sql
SELECT Region, City FROM Customer
LIMIT 10
```

```python
%%sql
SELECT City, Region FROM Employee
```

```python
%%sql
SELECT Region, City FROM Customer

INTERSECT 

SELECT Region, City FROM Employee
```

<h2>3. Except Operator</h2>

<img src="../media/Except.png" width="500"/>

```python
%%sql

SELECT * FROM Supplier LIMIT 3
```

```python
%%sql

SELECT * FROM Shipper

```

```python
%%sql

SELECT CompanyName FROM Supplier

EXCEPT

SELECT CompanyName FROM Shipper

LIMIT 10
```

<h2>Conclusion</h2>

Querying Data From multiple tables can be performed using either:
<ul>
    <li><b style="color:blue">JOIN Operators</b>
        <ul><li>(INNER) JOIN</li> <li>LEFT/RIGHT JOIN</li><li>FULL OUTER JOIN</li><li>CROSS JOIN</li></ul>
    </li>
    <li><b style="color:blue">SET Operators</b>
        <ul><li>UNION</li> <li>INTERSECT</li><li>EXCEPT</li></ul>
    </li>
</ul>

You will learn that <b style="color:blue">there is not a single way to fetch the same data</b> from a database! Just difference performance outcomes!
