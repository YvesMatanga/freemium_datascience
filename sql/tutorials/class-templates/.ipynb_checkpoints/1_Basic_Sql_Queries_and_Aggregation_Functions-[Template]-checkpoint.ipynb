<b>C. UPDATE Statements </b>

Modify a record or a set or records using a given criterion

```python
%%sql

UPDATE customer_list SET middle_name = 'ramaphosa' WHERE id=102
```

```python
%%sql
SELECT * FROM customer_list ORDER BY id DESC
```

<h4> D. DELETE Statements </h4>

remove a record or a set of records using a given criterion

<b>DELETE FROM table_name WHERE criterion</b>

```python
%%sql

DELETE FROM customer_list WHERE id = 102
```

```python
%%sql
SELECT * FROM customer_list ORDER BY id DESC
```

<h4>B. String Functions</h4>

<h5>LENGTH()</h5>

```python
%%sql

SELECT Title,  LENGTH(Title) AS 'number_of_characters' FROM albums LIMIT 10
```

<h5>UPPER(string)</h5>

```python
%%sql

SELECT Title, UPPER(Title) AS 'upper_case', LOWER(Title) AS 'lower_case' FROM albums LIMIT 10
```

<h5>SUBSTRING(strm, index, length)</h5>

```python
%%sql

SELECT Title, SUBSTRING(Title, 1, 5) AS 'title_portion' FROM albums LIMIT 10
```

<h5>CONCATENATION: ||</h5>

```python
%%sql

SELECT FirstName, LastName,  (FirstName||' '||LastName) AS 'full name' FROM customers LIMIT 10
```

<h4>D. Conditional flow functions </h4>

<h5>IIF(test, true_outcome, false_outcome)</h5>

```python
%%sql
SELECT * FROM invoices LIMIT 10
```

```python
%%sql

SELECT Total, IIF(total < 10, 'cheap','expensive') AS 'purchase category' FROM invoices LIMIT 5
```

<h5>CASE (test, option1, option2,...default)</h5>

```python
%%sql

SELECT * FROm media_types
```

```python
%%sql

SELECT Name, 

CASE Name
   WHEN 'MPEG audio file' THEN 'MPEG'
   WHEN 'Protected AAC audio file' THEN 'PAAC'
   WHEN 'Protected MPEG-4 video file' THEN 'PMPEG-4'
   WHEN 'Purchased AAC audio file' THEN 'PAAC'
   ELSE 'AAC'
END AS 'Type'

,LENGTH(Name) AS 'name_length' FROM media_types
```

<h5>IFNULL(is_null_condition, value)</h5>

```python
%%sql

SELECT BillingState, IFNULL(BillingState,'Empty') FROM invoices LIMIT 10
```

<h5>NULLIF()</h5>

```python
%%sql 

SELECT NULLIF(5,5)
```
