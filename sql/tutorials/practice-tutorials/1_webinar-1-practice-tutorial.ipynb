<h4>Question QS1-4:  Display the content STUDENT table including their corresponding instructors</h4>
<ul>
    <li>Display the student name, birth_date, instructor name and email</li>
    <li>Display the records in ascending order of student ages.
</ul>

<details style="color:blue">
    <summary>Click to reveal the <b>answer<b/></summary>

  **Answer:**
   %%sql
SELECT student.name AS 'student name', birth_date AS 'birthdate', instructor.name AS 'instructor name',
instructor.email AS 'instructor email' FROM student 
JOIN instructor ON instructor.id = student.instructor_id
ORDER BY birthdate DESC
        
</details>

<h4>Question QS1-6:  Display students name, birth_date and age</h4>
<ul>
    <li>Write a SELECT statement to display a student name, birthdate and age</li>
    <li>Use a strftime function to convert extract time information:<b> strftime(format, date_value)</b></li>
    <li>e.g. strftime('%Y', 'now') extract current year</li>
</ul>

<details style="color:blue">
    <summary>Click to reveal the <b>answer<b/></summary>

  **Answer:**
  %%sql
SELECT 
    name, 
    birth_date, 
    (strftime('%Y', 'now') - strftime('%Y', birth_date)) - 
    (strftime('%m-%d', 'now') < strftime('%m-%d', birth_date)) AS age
FROM student;   
</details>

<h4>Question QS1-7:  Delete the records of all students under 25</h4>
<ul>
    <li>Delete records of students whose age is below 25</li>
</ul>

<details style="color:blue">
    <summary>Click to reveal the <b>answer<b/></summary>

  **Answer:**
  %%sql
DELETE FROM student
WHERE (strftime('%Y', 'now') - strftime('%Y', birth_date)) - 
      (strftime('%m-%d', 'now') < strftime('%m-%d', birth_date)) < 25;
</details>
